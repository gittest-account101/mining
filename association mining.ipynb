```python
import pandas as pd
import numpy as np
import requests
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

url="https://raw.githubusercontent.com/stedy/Machine-Learning-with-R-datasets/master/groceries.csv"

text=requests.get(url).text
transact = [line.split(",") for line in text.split("\n")]

data= pd.DataFrame(transact)
data.head()
```

```python
data = data.fillna('')

transactions = data.values.tolist()
transactions = [[item for item in row if item != ''] for row in transactions]

print(transactions[:5])
```

```python
te = TransactionEncoder()
te_data = te.fit(transactions).transform(transactions)
df = pd.DataFrame(te_data, columns=te.columns_)

print("one-hot encoded DataFrame:")
df.head()
```

```python
freuquent_itemsets = apriori(df, min_support=0.01, use_colnames=True)

print("Frequent Itemsets:")
freuquent_itemsets.sort_values(by='support', ascending=False).head(10)

```

```python
rules = association_rules(freuquent_itemsets, metric="confidence", min_threshold=0.1)
strong_rules =  rules[(rules['lift']>1) & (rules['confidence']>0.5)]


print('\nStrong Association Rules: ')

strong_rules.head(10)
```
