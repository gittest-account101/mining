```python
def is_perfect(numbers):
 sum_divisors = sum(i for i in range(1,numbers) if numbers % i == 0)
 return sum_divisors == numbers

num = 28
if is_perfect(num):
 print(f"{num} is perfect number.")
else:
 print(f"{num} is not perfect number.")

```
