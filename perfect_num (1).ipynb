### number is perfect if the digits sum upto 10 

1 ===> 19 

2 ===> 28 

3 ===> 37 

WAP for the user to enter a number, and the output should be a perfect number

```python
def add_dig(z): 
    y=0
    digits = [int(d) for d in str(z)]
    for i in digits:
        y+=i     
    return y
j= int(input("enter number"))
x=j
while True:
    x=add_dig(x)
    # Condition check at the end
    if x<10:
        break
digit_need=10-x
print((j*10)+digit_need)
   

```

### array of numbers [9,11,8,5,7,10]
stock prices in chronological order of a company
function to calculate maximum profit

```python
x=[9,11,8,5,7,10]
max_x,max_y,pf=0,0,0

for i in x:
    for j in x[1:]:
        if j-i>pf:
            max_x,max_y,pf=i,j,(j-i)
    x=x[1:]
print(f"invest at:{max_x}\n withdraw at:{max_y}\n max_profit: {pf}")
```
